```python
import pandas as pd
import numpy as np
import os
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
import folium
from sklearn.feature_selection import chi2 # chi-squared test
from sklearn.feature_selection import SelectKBest
from sklearn.preprocessing import LabelEncoder
import json
```

```python
df_cleaned_no_outliers = pd.read_csv('extreme_cleaned_train_listing.csv')
df_cleaned_test = pd.read_csv('cleaned_test_listing_ml.csv')
```

```python
df_cleaned_no_outliers.columns
```

```python
x_features = [
    'host_is_superhost',
    'neighbourhood_cleansed',
    'neighbourhood_group_cleansed',
    'latitude',
    'longitude',
    'property_type',
    'room_type',
    'accommodates',
    'bathrooms',
    'bedrooms',
    'beds',
    'minimum_nights',
    'availability_365',
    'instant_bookable'
]
```

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 8))

# Histogram
axes[0].hist(df_cleaned_no_outliers["price"], bins=50)
axes[0].set_title("Price Distribution")
axes[0].set_xlabel("Price")
axes[0].set_ylabel("Frequency")

# Boxplot
axes[1].boxplot(df_cleaned_no_outliers["price"].dropna(), vert=False)
axes[1].set_title("Price Boxplot")
axes[1].set_xlabel("Price")

plt.tight_layout()
plt.show()
```

```python
features = [feature for feature in x_features if feature not in ['latitude', 'longitude']]

X = df_cleaned_no_outliers[features]
y = df_cleaned_no_outliers['price']

numeric_features = X.select_dtypes(include='number').columns


fig, axes = plt.subplots(2, 3, figsize=(16, 8))
axes = axes.flatten()

for ax, feature in zip(axes, numeric_features):
    ax.scatter(X[feature], y, alpha=0.3)
    ax.set_xlabel(feature)
    ax.set_ylabel("price")
    ax.set_title(f"{feature} vs price")


plt.tight_layout()
plt.show()
```

```python
# Rank room types by median price (highest to lowest)
room_order = (
    df_cleaned_no_outliers.groupby('room_type')['price']
    .median()
    .sort_values(ascending=False)
    .index
)

plt.figure(figsize=(10, 6))

sns.boxplot(
    x='room_type',
    y='price',
    data=df_cleaned_no_outliers,
    order=room_order,
    hue='room_type'     # one color is sufficient
)

plt.title('Price Distribution by Room Type')
plt.xlabel('Room Type', fontsize=12)
plt.ylabel('Price ($)', fontsize=12)

plt.show()
```

```python
room_type_counts = df_cleaned_no_outliers["room_type"].value_counts()

plt.figure(figsize=(8, 5))

bars = plt.bar(
    room_type_counts.index,
    room_type_counts.values
)

plt.xlabel("Room Type")
plt.ylabel("Number of Listings")
plt.title("Distribution of Room Types")

# Add values on top of bars
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{int(bar.get_height()):,}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python

initial_available_columns = [
    'host_is_superhost',
    'neighbourhood_cleansed',
    'neighbourhood_group_cleansed',
    'latitude',
    'longitude',
    'property_type',
    'room_type',
    'accommodates',
    'bathrooms',
    'bedrooms',
    'beds',
    'price',
    'minimum_nights',
    'price_log',
    'availability_365',
    'instant_bookable'
]

# Create a copy
df_extreme_cleaned_ml = df_cleaned_no_outliers.copy()

# Drop unnecessary columns
df_extreme_cleaned_ml = df_extreme_cleaned_ml[initial_available_columns]

# -----------------------------------------
# Binary Encode
# -----------------------------------------
df_extreme_cleaned_ml["instant_bookable"] = (
    df_extreme_cleaned_ml["instant_bookable"].map({
        'f': 0,
        't': 1
    })
)

df_extreme_cleaned_ml["host_is_superhost"] = (
    df_extreme_cleaned_ml["host_is_superhost"].map({
        'f': 0,
        't': 1
    })
)

# -----------------------------------------
# One-Hot Encode
# -----------------------------------------
one_hot_cols = [
    "room_type",
    "neighbourhood_group_cleansed"
]

df_extreme_cleaned_ml = pd.get_dummies(
    df_extreme_cleaned_ml,
    columns=one_hot_cols,
    drop_first=False
)

# -----------------------------------------
# Label Encode
# -----------------------------------------
with open("encoder_maps.json", "r") as f:
    encoder_maps = json.load(f)

for col in ["property_type", "neighbourhood_cleansed"]:
    df_extreme_cleaned_ml[col] = df_extreme_cleaned_ml[col].map(encoder_maps[col])

# -----------------------------------------
# Check data types
# -----------------------------------------
print(df_extreme_cleaned_ml.dtypes)

# Correlation matrix
corr = df_extreme_cleaned_ml.corr(numeric_only=True)
```

```python
for col in ["property_type", "neighbourhood_cleansed"]:
    df1_categories = set(df_extreme_cleaned_ml[col].dropna().unique())
    df2_categories = set(df_cleaned_test[col].dropna().unique())

    unseen = df2_categories - df1_categories

    if unseen:
        print(f"{col}: ❌ Categories in df2 but not df1:")
        print(unseen)
    else:
        print(f"{col}: ✅ All df2 categories exist in df1")
```

```python
plt.figure(figsize=(20, 16))

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.2
)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```
